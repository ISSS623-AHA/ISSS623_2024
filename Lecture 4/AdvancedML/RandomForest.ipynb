```python
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier(
    # 1. Bagging: build many trees
    n_estimators=200,        # number of trees in the forest
    bootstrap=True,          # sample rows with replacement
    max_samples=None,        # each tree gets bootstrap sample of size n_samples

    # 2. Random feature selection at each split
    max_features="sqrt",     # consider sqrt(p) features at each split

    # 3. Out-of-bag error estimation
    oob_score=True,          # estimate performance using OOB samples

    # 4. Model complexity restrictions
    max_depth=None,          # None = grow trees until stopping criteria are met
    min_samples_split=2,     # minimum samples needed to split a node
    min_samples_leaf=1,      # minimum samples required in a leaf node
    max_leaf_nodes=None,     # None = unlimited number of leaf nodes
    ccp_alpha=0.0,           # 0.0 = no cost-complexity pruning

    # Practical settings
    random_state=42,         # reproducible results
    n_jobs=-1                # use all available CPU cores
)
```
