```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Mock patient-level hospital data
data = pd.DataFrame({
    "age": [45, 50, 62, 70, 58, 80, 66, 75],
    "comorbidity_count": [1, 2, 3, 4, 2, 5, 3, 4],
    "prior_admissions": [0, 1, 1, 2, 1, 3, 2, 2],
    "length_of_stay": [3.2, 4.1, 5.8, 7.0, 4.9, 8.6, 6.5, 7.5]
})

# Predictors
X = data[["age", "comorbidity_count", "prior_admissions"]]

# Outcome
y = data["length_of_stay"]

# OLS model with intercept
ols_model = LinearRegression(fit_intercept=True)

# Fit model
ols_model.fit(X, y)

# Print intercept and coefficients
print("Intercept:", ols_model.intercept_)
print("Coefficients:", ols_model.coef_)

# Predictions
y_pred = ols_model.predict(X)

# Simple model performance
print("MSE:", mean_squared_error(y, y_pred))
print("R-squared:", r2_score(y, y_pred))
```
